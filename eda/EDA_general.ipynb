# SOBUSA | Completitud de conteos por periodo

Edición pública con datos exclusivamente sintéticos. Conserva el procedimiento indicado del EDA existente. Los resultados son ilustrativos.

```python
import pandas as pd
import matplotlib.pyplot as plt

conteos8 = pd.DataFrame({
    "idprogramacion": range(12),
    "Mes": pd.date_range("2030-01-01", periods=12, freq="9D").to_period("M"),
    "Sin pasajeros": [i % 5 == 0 for i in range(12)],
    "Sin bajadas": [i % 4 == 0 for i in range(12)],
})
conteos8["Algún conteo ausente"] = conteos8["Sin pasajeros"] | conteos8["Sin bajadas"]
```

```python
def resumen_conteos8(campo):
    resumen = conteos8.groupby(campo, dropna=False).agg(
        Finalizadas=("idprogramacion", "size"),
        Sin_pasajeros=("Sin pasajeros", "sum"),
        Sin_bajadas=("Sin bajadas", "sum"),
        Algun_conteo_ausente=("Algún conteo ausente", "sum")
    ).reset_index()
    resumen["% con algún conteo ausente"] = (
        100 * resumen["Algun_conteo_ausente"] / resumen["Finalizadas"]
    ).round(2)
    return resumen
```

```python
mensual = resumen_conteos8("Mes")
display(mensual)
mensual.plot.bar(x="Mes", y="% con algún conteo ausente", title="Ejemplo sintético de completitud", rot=0)
plt.show()
```
