# Ruta 21 | Cobertura y faltantes por estado

Edición pública con datos exclusivamente sintéticos. Conserva el procedimiento indicado del EDA existente. Los resultados son ilustrativos.

```python
import pandas as pd
import matplotlib.pyplot as plt

p = pd.DataFrame({
    "finalizado": ["S", "S", "N", "S", "S", "N"],
    "cancelada": ["N", "N", "S", "N", "N", "S"],
    "fechasalida": pd.date_range("2030-01-01", periods=6, freq="h"),
    "fechainicio": pd.to_datetime(["2030-01-01 00:05", None, None, "2030-01-01 03:05", None, None]),
    "fechafin": pd.to_datetime(["2030-01-01 01:05", None, None, "2030-01-01 04:05", None, None]),
    "numeropasajeros": [22, None, None, 18, 12, None],
    "numerobajadas": [22, None, None, 18, None, None],
})
```

```python
est=p[['finalizado','cancelada']].copy()
for col in ['fechasalida','fechainicio','fechafin','numeropasajeros','numerobajadas']:est[col+'_ausente']=p[col].isna()
estados=est.groupby(['finalizado','cancelada'],dropna=False).agg(denominador=('fechafin_ausente','size'),
    inicio_ausente=('fechainicio_ausente','sum'),fin_ausente=('fechafin_ausente','sum'),
    pasajeros_ausente=('numeropasajeros_ausente','sum'),bajadas_ausente=('numerobajadas_ausente','sum')).reset_index()
for col in ['inicio_ausente','fin_ausente','pasajeros_ausente','bajadas_ausente']:estados[col+'_pct']=100*estados[col]/estados.denominador
```

```python
display(estados)
estados.set_index(["finalizado", "cancelada"])[["pasajeros_ausente_pct", "bajadas_ausente_pct"]].plot.bar(
    title="Ruta 21 · porcentaje ausente en datos ficticios", ylabel="Porcentaje", rot=0)
plt.show()
```
